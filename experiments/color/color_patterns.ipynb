# Mining Color Patterns in the Mucha Corpus

This notebook is the reading interface for `mucha_color_patterns_v1`. The reproducible implementation lives in `run_experiment.py`, and the committed outputs allow the analysis to be inspected without repeating palette extraction.

```python
from pathlib import Path
import json
from IPython.display import Image, display

HERE = Path.cwd()
if HERE.name != 'color':
    HERE = HERE / 'experiments' / 'color'
OUTPUTS = HERE / 'outputs'
FORCE_RERUN = False
```

## Part 1 — Reading a Single Image with Pylette

A59 compares KMeans/8 with OKLab palettes of 8, 16, and 32 colors. Eight colors describe large-area tonality; 16 and 32 increasingly retain secondary and accent colors.

```python
display(Image(filename=str(OUTPUTS / 'figures' / 'a59_palette_comparison.png')))
```

## Part 2 — Extracting Multi-scale Palettes from the Corpus

The frozen input contains all catalogue-designated color plates. Extraction produces three combined raw files plus normalized long-form swatches and auditable run/failure records.

```python
if FORCE_RERUN:
    import subprocess, sys
    subprocess.run([sys.executable, str(HERE / 'run_experiment.py'), 'extract', '--force'], check=True)
summary = json.loads((OUTPUTS / 'mining' / 'analysis_summary.json').read_text(encoding='utf-8'))
summary['summary']
```
